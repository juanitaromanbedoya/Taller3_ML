```python
import pandas as pd 
import requests

url = "https://rickandmortyapi.com/api/character"
data = requests.get(url).json()
data_api = pd.DataFrame(data['results'])

data_api = data_api[['id', 'name', 'status', 'species', 'type', 'gender', 'origin', 'location', 'image', 'episode', 'url', 'created']]

print (data_api.head())
```
